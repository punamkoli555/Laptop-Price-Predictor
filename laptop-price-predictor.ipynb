```python
import pandas as pd
```

```python
df = pd.read_excel('/kaggle/input/laptop-specifications-and-pricing-dataset/Cleaned_dataset.xlsx')

df.head()
```

```python
df['Ram'] = df['Ram'].str.replace('GB', '')

df['Weight'] = df['Weight'].str.replace('kg', '')

df['Ram'] = df['Ram'].astype('int32')

df['Weight'] = df['Weight'].astype('float32')

df.head()
```

```python
df['Touchscreen'] = df['ScreenResolution'].apply(lambda x: 1 if 'Touchscreen' in x else 0)
df['Ips'] = df['ScreenResolution'].apply(lambda x: 1 if 'IPS' in x else 0)

df.head()
```

```python
new = df['ScreenResolution'].str.split('x', n=1, expand=True)

df['X_res'] = new[0]
df['Y_res'] = new[1]

df['X_res'] = df['X_res'].str.replace(',', '')
df['X_res'] = df['X_res'].str.findall(r'(\d+.?\d+)').apply(lambda x: x[0]).astype('int')

df['Y_res'] = df['Y_res'].astype('int')

df['ppi'] = ((df['X_res'] ** 2 + df['Y_res'] ** 2) ** 0.5) / df['Inches']
df['ppi'] = df['ppi'].astype('float')

df.drop(columns=['Inches', 'X_res', 'Y_res', 'ScreenResolution'], inplace=True)

df
```

```python
df['Cpu Name'] = df['Cpu'].apply(lambda x: " ".join(x.split()[0:3]))

def fetch_processor(cpu_name):
    if 'Intel Core i7' in cpu_name:
        return 'Intel Core i7'
    elif 'Intel Core i5' in cpu_name:
        return 'Intel Core i5'
    elif 'Intel Core i3' in cpu_name:
        return 'Intel Core i3'
    elif 'Intel' in cpu_name:
        return 'Other Intel Processor'
    elif 'AMD' in cpu_name:
        return 'AMD Processor'
    else:
        return 'Other Processor'

df['Cpu brand'] = df['Cpu Name'].apply(fetch_processor)

df.drop(columns=['Cpu', 'Cpu Name'], inplace=True)

df.head()
```

```python
def cat_os(inp):
    if inp in ['Windows 10', 'Windows 7', 'Windows 10 S']:
        return 'Windows'
    elif inp in ['macOS', 'Mac OS X']:
        return 'Mac'
    else:
        return 'Others/No OS/Linux'

df['os'] = df['OpSys'].apply(cat_os)

df.drop(columns=['OpSys'], inplace=True)

df.head()
```

```python
df.dtypes
```

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

# Assuming 'df' is your DataFrame with the specified columns
categorical_columns = ['Company', 'TypeName', 'Gpu brand', 'Cpu brand', 'os']
numerical_columns = ['Ram', 'HDD', 'SSD', 'Weight', 'Price', 'Touchscreen', 'Ips', 'ppi']

# Split the DataFrame into numerical and categorical parts
numerical_df = df[numerical_columns]
categorical_df = df[categorical_columns]

# Apply one-hot encoding to the categorical columns
encoder = OneHotEncoder(drop='first')
encoded_array = encoder.fit_transform(categorical_df).toarray()
encoded_df = pd.DataFrame(encoded_array, columns=encoder.get_feature_names_out(categorical_columns))

# Combine the numerical and encoded categorical DataFrames
merged_df = pd.concat([encoded_df, numerical_df], axis=1)

merged_df.head()
```

```python
from sklearn.model_selection import train_test_split
import numpy as np

X = merged_df.drop(columns=['Price'])
y = np.log(merged_df['Price'])

X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.85, test_size=0.15, random_state=2)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import r2_score, mean_absolute_error

step1 = ColumnTransformer(transformers=[
    ('encoder', OneHotEncoder(drop='first'), [0, 1, 6, 10, 11])
], remainder='passthrough')

step2 = LinearRegression()

pipe = Pipeline(steps=[
    ('preprocessor', step1),
    ('regressor', step2)
])

pipe.fit(X_train, y_train)

y_pred = pipe.predict(X_test)

score_r2 = r2_score(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)

print("R-squared Score:", score_r2)
print("Mean Absolute Error:", mae)
```

```python
from sklearn.linear_model import Ridge


step1 = ColumnTransformer(transformers=[
    ('encoder', OneHotEncoder(drop='first'), [0, 1, 6, 10, 11])
], remainder='passthrough')

step2 = Ridge(alpha=10)

pipe = Pipeline(steps=[
    ('preprocessor', step1),
    ('regressor', step2)
])

pipe.fit(X_train, y_train)

y_pred = pipe.predict(X_test)

score_r2 = r2_score(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)

print("R-squared Score:", score_r2)
print("Mean Absolute Error:", mae)
```
